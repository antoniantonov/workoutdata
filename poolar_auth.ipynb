```python
## POST to Pola Polar endpoint to obtain a access token
## To get auth code invoke this from the browser: https://flow.polar.com/oauth2/authorization?response_type=code&client_id=fddbcde4-65b8-4e25-8c74-08cab0a76e82
## Here is the auth code: https://humandcoded.com/?code=6a61a6b64d94927dd9b27662e3690903 

import requests
import base64

def get_polar_access_token(client_id, client_secret, auth_code, redirect_uri=None):
    """
    Get access token from Polar API using authorization code.
    
    Args:
        client_id (str): Your Polar API client ID
        client_secret (str): Your Polar API client secret
        auth_code (str): Authorization code received from the authorization endpoint
        redirect_uri (str, optional): Must match the redirect_uri used in authorization step
        
    Returns:
        dict: JSON response containing access token and other details
    """
    # Build the endpoint URL
    url = "https://polarremote.com/v2/oauth2/token"
    
    # Create the Authorization header with base64 encoded client_id:client_secret
    auth_string = f"{client_id}:{client_secret}"
    encoded_auth = base64.b64encode(auth_string.encode()).decode()
    
    # Set up the headers
    headers = {
        "Authorization": f"Basic {encoded_auth}",
        "Content-Type": "application/x-www-form-urlencoded",
        "Accept": "application/json;charset=UTF-8"
    }
    
    # Prepare the request data
    data = {
        "grant_type": "authorization_code",
        "code": auth_code
    }
    
    # Add redirect_uri if provided
    if redirect_uri:
        data["redirect_uri"] = redirect_uri
    
    # Make the POST request
    response = requests.post(url, headers=headers, data=data)
    
    # Check if the request was successful
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error: {response.status_code}")
        print(response.text)
        return None

# Example usage:
client_id = "your_client_id"
client_secret = "your_client_secret"
auth_code = "authorization_code_from_previous_step"
redirect_uri = "https://your-redirect-uri.com/callback"  # Optional, if used in authorization step

token_response = get_polar_access_token(client_id, client_secret, auth_code, redirect_uri)
if token_response:
    access_token = token_response.get("access_token")
    refresh_token = token_response.get("refresh_token")
    print(f"Access token: {access_token}")
    print(f"Refresh token: {refresh_token}")

"""
curl -X POST https://polarremote.com/v2/oauth2/token \
  -H 'Content-Type: application/x-www-form-urlencoded' \
  -H 'Accept: application/json;charset=UTF-8' \
  -H 'Authorization: Basic ZmRkYmNkZTQtNjViOC00ZTI1LThjNzQtMDhjYWIwYTc2ZTgyOjZjM2NiYTEyLTNiODQtNGJkZi05NGQ2LTZlZDY1Mjc2MjU2Mw==' \
  --data-urlencode "grant_type=authorization_code&code=c5d1f52b95e1ae6d6512ba29abb69900"
"""
```
